```python
import pandas as pd

model_comparison_time = pd.DataFrame({
    "Model": ["Random Forest", "XGBoost"],
    "Precision": [0.999529, 0.979330],
    "Recall": [0.998353, 0.735765],
    "F1": [0.998941, 0.840253],
    "ROC-AUC": [0.9999998, 0.986979],
    "PR-AUC": [0.999945, 0.940951]
})

model_comparison_time
```

```python
'''The comparison is now on the same time-based holdout set,
 so this is a much more meaningful evaluation than the earlier random split.

What the table shows
Metric	Random Forest	XGBoost
Precision	99.95%	97.93%
Recall	99.84%	73.58%
F1	99.89%	84.03%
ROC-AUC	~100%	98.70%
PR-AUC	99.9945%	94.10%

The next thing we should do is visualize this comparison so the difference
 is easy to understand and useful for your project report.'''
```

```python
import matplotlib.pyplot as plt

metrics = ["Precision", "Recall", "F1", "ROC-AUC", "PR-AUC"]

rf_scores = model_comparison_time.loc[0, metrics].values
xgb_scores = model_comparison_time.loc[1, metrics].values

x = range(len(metrics))

plt.figure(figsize=(10, 6))

plt.bar(
    [i - 0.2 for i in x],
    rf_scores,
    width=0.4,
    label="Random Forest"
)

plt.bar(
    [i + 0.2 for i in x],
    xgb_scores,
    width=0.4,
    label="XGBoost"
)

plt.xticks(x, metrics)
plt.ylabel("Score")
plt.ylim(0, 1.05)
plt.title("Time-Based Model Comparison")
plt.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

# Random Forest confusion matrix
rf_cm = np.array([
    [1244484, 2],
    [7, 4243]
])

# XGBoost confusion matrix
xgb_cm = np.array([
    [1244420, 66],
    [1123, 3127]
])


```

```python
ConfusionMatrixDisplay(
    confusion_matrix=rf_cm,
    display_labels=["Normal", "Fraud"]
).plot()

plt.title("Random Forest - Time-Based Confusion Matrix")
plt.show()
```

```python
ConfusionMatrixDisplay(
    confusion_matrix=xgb_cm,
    display_labels=["Normal", "Fraud"]
).plot()

plt.title("XGBoost - Time-Based Confusion Matrix")
plt.show()
```
